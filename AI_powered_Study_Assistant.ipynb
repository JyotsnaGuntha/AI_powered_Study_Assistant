Import Module from Gemini Package

```python
from google import genai

# Import the user's data from colab to our notebook so we can access the secrets
from google.colab import userdata
```

Use Client class to create an instance (for interacting with Gemini API)

```python
#The client object provides methods and attributes for API interaction, such as
 #client.models
 #client.auth
 #client.config

#Instead of directly pasting the API Key here, we'll fetch the API key from userdata.
client = genai.Client(api_key = userdata.get("GEMINI_API_KEY_AI_Study_Assistant"))
print(client)
```

Use client object for communicating to Gemini LLM

```python
def study_assistant(user_prompt):
  response = client.models.generate_content( #allows us to send prompt to the AI model and get a response
      model="gemini-2.5-flash", #model to use while generating content
      contents=user_prompt #The actual prompt that will serve as the input to our LLM
  )
  return response
```

API key Auth -> Request sent to LLM -> Response Returned -> Stored in response variable -> Printed with print(output)

```python
output = study_assistant("Explain LLM Architecture")
```
